```python
import matplotlib.pyplot as plt
time = [1, 2, 3, 4, 5]

altitudes = np.array([400, 420, 450, 480, 500])

plt.plot(time, altitudes, marker='o')

plt.title("Satellite Altitude vs Time")
plt.xlabel("Time (Hours)")
plt.ylabel("Altitude (km)")

plt.grid(True)

plt.show()
```

```python
time = [1, 2, 3, 4, 5]

actual_altitude = [400, 420, 450, 480, 500]
target_altitude = [410, 430, 450, 470, 490]

plt.plot(time, actual_altitude, marker='o', label='Actual Altitude')
plt.plot(time, target_altitude, marker='s', label='Target Altitude')

plt.title("Satellite Altitude Comparison")
plt.xlabel("Time (Hours)")
plt.ylabel("Altitude (km)")

plt.legend()
plt.grid(True)

plt.show()
```

```python
from scipy.optimize import root_scalar

def equation(x):
    return x**2 - 4

solution = root_scalar(equation, bracket=[1, 3])

print("Solution:", solution.root)
```

```python
# negative_solution = root_scalar(equation, bracket=[-3, -1])

# print("Negative solution:", negative_solution.root)


def equation(x):
    return x**2 - 25

solution = root_scalar(equation, bracket=[4, 6])

print("Positive root:", solution.root)

def equation(x):
    return x**2 - 9

solution = root_scalar(equation, bracket=[-4, -2])

print("Negative root:", solution.root)
```

```python
def engineering_equation(x):
    return x**2 + 3*x - 10

solution = root_scalar(
    engineering_equation,
    bracket=[1, 3]
)

print("Positive root:", solution.root)
```

```python
negative_solution = root_scalar(
    engineering_equation,
    bracket=[-5, -2]
)

print("Negative root:", negative_solution.root)
```

```python
def equation(x):
    return x**2 + 2*x - 15

solution = root_scalar(equation, bracket=[-6, -4])

print("Negative root:", solution.root)
```

```python
def satellite_equation(Pt):
    return Pt + 8 + 5 - 30 - (-10)

solution = root_scalar(
    satellite_equation,
    bracket=[0, 15]
)

print("Required transmit power:", solution.root, "dBm")
```
