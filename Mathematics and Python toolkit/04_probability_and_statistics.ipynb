```python
import numpy as np

rng = np.random.default_rng(42)

# Probability that one packet succeeds
p_success = 0.9

# TODO 1: Simulate 1000 transmissions
# Each result should be 1 for success or 0 for failure
results = rng.choice([0, 1], size=1000, p=[1 - p_success, p_success])

# TODO 2: Count successful transmissions
successes = np.sum(results)

# TODO 3: Calculate the observed success rate
success_rate = successes / len(results)

print("Total transmissions:", len(results))
print("Successful transmissions:", successes)
print("Observed success rate:", success_rate)
print("Observed success percentage:", success_rate * 100, "%")
```

```python
# TODO 1: Count failed transmissions
failures = len(results) - successes

# TODO 2: Calculate the observed failure rate
failure_rate = failures / len(results)

# TODO 3: Calculate the theoretical number of failures
expected_failures = len(results) * (1 - p_success)

print("Failed transmissions:", failures)
print("Observed failure rate:", failure_rate)
print("Expected failures:", expected_failures)
```

Question 2 of 6 — Poisson distribution
A satellite memory module experiences an average of 2 radiation-induced bit upsets per day.

Calculate:
1. Probability of zero upsets in one day.
2. Probability of five or more upsets in one day.

```python

import math

lam = 2

# Probability of zero upsets
p_zero = math.exp(-lam) * lam**0 / math.factorial(0)

# Probability of five or more upsets
p_zero_to_four = 0

for k in range(5):
    p = math.exp(-lam) * lam**k / math.factorial(k)
    p_zero_to_four += p

p_five_or_more = 1 - p_zero_to_four

print("Probability of zero upsets:", p_zero)
print("Probability of five or more:", p_five_or_more)

print("Zero upsets (%):", p_zero * 100)
print("Five or more (%):", p_five_or_more * 100)
```

Question 3 of 6 — Normal distribution
A satellite temperature sensor has:
- Mean temperature = 30°C
- Standard deviation = 3°C
Calculate the temperature ranges containing approximately:
1. 68% of measurements.
2. 95% of measurements.
3. 99.7% of measurements.

```python

mean_temp = 30
std_temp = 3

# 68% range (1 standard deviation)
lower_68 = mean_temp - std_temp
upper_68 = mean_temp + std_temp

# 95% range (2 standard deviations)
lower_95 = mean_temp - 2 * std_temp
upper_95 = mean_temp + 2 * std_temp

# 99.7% range (3 standard deviations)
lower_997 = mean_temp - 3 * std_temp
upper_997 = mean_temp + 3 * std_temp

print("68% range:", lower_68, "to", upper_68)
print("95% range:", lower_95, "to", upper_95)
print("99.7% range:", lower_997, "to", upper_997)
```

Question 4 of 6 — Exponential distribution
A satellite communication component has a failure rate of:
\[
\lambda=0.2\text{ failures/hour}
\]

Calculate:
1. Probability of surviving beyond 3 hours.
2. Probability of failure within 3 hours.
3. Mean Time Between Failures (MTBF).

```python

import math

lam = 0.2
t = 3

# Probability of surviving beyond 3 hours
survival_probability = math.exp(-lam * t)

# Probability of failure within 3 hours
failure_probability = 1 - survival_probability

# Mean Time Between Failures
mtbf = 1 / lam

print("Survival probability:", survival_probability)
print("Failure probability:", failure_probability)
print("MTBF:", mtbf, "hours")
```

Question 5 of 6 — Failure rate and reliability
A satellite battery has an MTBF of 200 hours.

Calculate:
1. Failure rate (\(\lambda\)).
2. Reliability after a 100-hour mission.

```python
import math

mtbf = 200
mission_time = 100

# Failure rate
failure_rate = 1 / mtbf

# Reliability after 100 hours
reliability = math.exp(-failure_rate * mission_time)

print("Failure rate:", failure_rate)
print("Reliability:", reliability)
```

Question 6 of 6 — Confidence interval
A satellite sensor records temperature measurements:
- Sample mean = 50°C
- Sample standard deviation = 4°C
- Sample size = 400
- 95% confidence multiplier = 1.96
Calculate the 95% confidence interval.

```python
mean = 50
std = 4
n = 400
z = 1.96

# Standard error
standard_error = std / np.sqrt(n)

# Margin of error
margin = z * standard_error

# Confidence interval
lower = mean - margin
upper = mean + margin

print("Standard error:", standard_error)
print("Margin of error:", margin)
print("95% CI:", lower, "to", upper)
```
