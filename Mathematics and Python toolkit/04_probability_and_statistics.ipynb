```python
import numpy as np

rng = np.random.default_rng(42)

# Probability that one packet succeeds
p_success = 0.9

# TODO 1: Simulate 1000 transmissions
# Each result should be 1 for success or 0 for failure
results = rng.choice([0, 1], size=1000, p=[1 - p_success, p_success])

# TODO 2: Count successful transmissions
successes = np.sum(results)

# TODO 3: Calculate the observed success rate
success_rate = successes / len(results)

print("Total transmissions:", len(results))
print("Successful transmissions:", successes)
print("Observed success rate:", success_rate)
print("Observed success percentage:", success_rate * 100, "%")
```

```python
# TODO 1: Count failed transmissions
failures = len(results) - successes

# TODO 2: Calculate the observed failure rate
failure_rate = failures / len(results)

# TODO 3: Calculate the theoretical number of failures
expected_failures = len(results) * (1 - p_success)

print("Failed transmissions:", failures)
print("Observed failure rate:", failure_rate)
print("Expected failures:", expected_failures)
```

```python
rng = np.random.default_rng(42)

results = rng.binomial(
    n=1,
    p=p_success,
    size=20
)

print("Transmission results:")
print(results)

successes = np.sum(results)
failures = len(results) - successes

print("Successes:", successes)
print("Failures:", failures)
```

```python

import numpy as np

rng = np.random.default_rng(42)

p_success = 0.9

# Simulate 100 transmissions
results = rng.binomial(
    n=5,
    p=p_success,
    size=100
)

# Count successful transmissions
successes = np.sum(results)

failures = len(results) - successes

print("Total transmissions:", len(results))
print("Successes:", successes)
print("Failures:", failures)
print("Observed success rate:", successes / len(results))
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

# Simulated satellite temperature measurements
mean_temp = 20.0
std_temp = 2.0
measurements = rng.normal(
    loc=mean_temp,
    scale=std_temp,
    size=1000
)

print("Measured average:", np.mean(measurements))
print("Measured standard deviation:", np.std(measurements))

plt.hist(measurements, bins=30, edgecolor="black")
plt.xlabel("Temperature (°C)")
plt.ylabel("Number of measurements")
plt.title("Satellite Sensor: Normal Distribution")
plt.show()
```

```python
import numpy as np

rng = np.random.default_rng(42)

mean_temp = 30
std_temp = 3

measurements = rng.normal(
    loc=mean_temp,
    scale=std_temp,
    size=10_000
)

within_1_sigma = np.mean(
    (measurements >= mean_temp - std_temp) &
    (measurements <= mean_temp + std_temp)
)

within_2_sigma = np.mean(
    (measurements >= mean_temp - 2 * std_temp) &
    (measurements <= mean_temp + 2 * std_temp)
)

within_3_sigma = np.mean(
    (measurements >= mean_temp - 3 * std_temp) &
    (measurements <= mean_temp + 3 * std_temp)
)

print("Within 1 sigma:", within_1_sigma * 100, "%")
print("Within 2 sigma:", within_2_sigma * 100, "%")
print("Within 3 sigma:", within_3_sigma * 100, "%")
```
