```python

import numpy as np
import matplotlib.pyplot as plt

# Initial conditions
g = 9.81       # gravity (m/s²)
h0 = 100       # initial height (m)
v0 = 0         # initial velocity (m/s)
dt = 0.01      # time step (s) 

# Numerical integration
t = np.arange(0, 4.6, dt)
h = np.zeros(len(t))
v = np.zeros(len(t))

h[0] = h0
v[0] = v0

for i in range(len(t) - 1):
    h[i + 1] = h[i] + v[i] * dt
    v[i + 1] = v[i] - g * dt

# Exact solution
h_exact = h0 + v0 * t - 0.5 * g * t**2

# Compare the results
print("Numerical height at 2 s:", h[np.argmin(abs(t - 2))], "m")
print("Exact height at 2 s:", h_exact[np.argmin(abs(t - 2))], "m")

plt.plot(t, h, label="Numerical integration")
plt.plot(t, h_exact, "--", label="Exact formula")
plt.xlabel("Time (seconds)")
plt.ylabel("Height (metres)")
plt.title("Object Falling Under Gravity")
plt.legend()
plt.grid(True)
plt.show()
```


## Experiment 02 — Time-Step Accuracy

**Objective:** Understand how the time step affects numerical integration accuracy.

**Experiment:**
- Initial height: 100 m
- Initial velocity: 0 m/s
- Gravity: 9.81 m/s²
- First time step: 0.1 s
- Second time step: 0.01 s

**Observation:** The numerical curve gets closer to the exact solution when the time step is reduced.

**Conclusion:** Smaller time steps generally improve accuracy for this numerical method, at the cost of more computation.

```python

# Compare numerical and exact height at 2 seconds
index = np.argmin(np.abs(t - 2))

numerical_height = h[index]
exact_height = h_exact[index]

error = abs(numerical_height - exact_height)

print("Time:", t[index], "seconds")
print("Numerical height:", round(numerical_height, 4), "m")
print("Exact height:", round(exact_height, 4), "m")
print("Absolute error:", round(error, 4), "m")
```

```python
g = 9.81
h = 100.0
v = 0.0
dt = 0.01
total_time = 2.0

steps = int(total_time / dt)

for i in range(steps):
    # TODO 1: Update height using the current velocity
    h += v * dt
    # TODO 2: Update velocity using gravity
    v -= g * dt

print("Simulated height:", h, "m")
print("Simulated velocity:", v, "m/s")

exact_height = 100 - 0.5 * 9.81 * 2**2
error = abs(exact_height - h)

print("Exact height:", exact_height, "m")
print("Numerical height:", h, "m")
print("Absolute error:", error, "m")
```
