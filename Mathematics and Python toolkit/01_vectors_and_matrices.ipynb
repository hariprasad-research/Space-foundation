## Vectors, Cross Products, Coordinate Frames and Matrices

Objective:
Represent mathematical concepts, calculate results by hand,
and verify them using NumPy.

Applications:
- Satellite position and velocity
- Angular momentum
- Coordinate transformations
- Satellite pointing

```python
import numpy as np

theta = np.radians(90)

R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
])

v_B = np.array([1, 0, 0])

v_A = R @ v_B

print("Rotation matrix:")
print(np.round(R, 2))

print("Original vector:", v_B)
print("Rotated vector:", np.round(v_A, 2))
```

```python

# Rotate a vector 90 degrees about the z-axis
theta = np.radians(90)

R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
])

v_B = np.array([1, 0, 0])

v_A = R @ v_B

print("Original coordinates:", v_B)
print("Transformed coordinates:", np.round(v_A, 6))
```


## What I learned

1. Dot product produces a scalar.
2. Cross product produces a perpendicular vector.
3. Coordinate frames describe quantities relative to chosen axes.
4. Rotation matrices transform vector coordinates.
5. Matrix multiplication combines mathematical transformations.
